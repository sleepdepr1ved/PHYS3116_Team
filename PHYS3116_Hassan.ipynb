```python
import pandas as pd
import matplotlib.pyplot as plt

harris = pd.read_csv(r"C:\Users\HassanNoun\OneDrive - UNSW\Desktop\PHYS3116\Group Assessment\GitHubClone_PHYS3116\PHYS3116_Team\HarrisPartI.csv")

# Extracting key spatial coordinates
R = harris["R_gc"]
X = harris["X"]
Y = harris["Y"]
Z = harris["Z"]

# Create subplots
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

# R vs |Z|
axes[0].scatter(R, abs(Z), c='mediumpurple', alpha=0.6, s=60)
axes[0].set_xlabel(r"$R_{gc}$ (kpc)", fontsize=12)
axes[0].set_ylabel(r"$|z|$ (kpc)", fontsize=12)

# X vs Z
axes[1].scatter(X, Z, c='mediumpurple', alpha=0.6, s=60)
axes[1].set_xlabel(r"$x$ (kpc)", fontsize=12)
axes[1].set_ylabel(r"$z$ (kpc)", fontsize=12)

# X vs Y
axes[2].scatter(X, Y, c='mediumpurple', alpha=0.6, s=60)
axes[2].set_xlabel(r"$x$ (kpc)", fontsize=12)
axes[2].set_ylabel(r"$y$ (kpc)", fontsize=12)

plt.tight_layout()
plt.show()
```

$v_\text{LSR}$ is the radial velocity of a cluster corrected for the motion of the Sun around the Milky Way, expressed relative to the local standard of rest (LSR). It represents how fast the cluster is moving toward or away from the Galaxy's average rotation frame.

```python
harris3 = pd.read_csv(r"C:\Users\HassanNoun\OneDrive - UNSW\Desktop\PHYS3116\Group Assessment\GitHubClone_PHYS3116\PHYS3116_Team\HarrisPartIII.csv")

# Extract columns
v_LSR = harris3["v_LSR"]

# Simple scatter plot
plt.figure(figsize=(6, 4))
plt.scatter(R, v_LSR, c='mediumpurple', alpha=0.7, s=60)

plt.xlabel(r"$R_{gc}$ (kpc)", fontsize=12)
plt.ylabel(r"$v_{\mathrm{LSR}}$ (km/s)", fontsize=12)
plt.title("Rotation of Globular Clusters (v$_{LSR}$ vs R$_{gc}$)", fontsize=13)
plt.axhline(0, color='gray', linestyle='--', linewidth=1)

plt.tight_layout()
plt.show()
```
